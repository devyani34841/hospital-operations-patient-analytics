```python
department = df["Department"].value_counts()

plt.figure(figsize=(5,3))
department.plot(kind="bar")
plt.title("Number of Patients by Department")
plt.xlabel("Department")
plt.ylabel("Number of Patients")
plt.show()
```

```python
payment = df["PaymentMode"].value_counts()

plt.figure(figsize=(7,7))
payment.plot(kind="pie", autopct="%1.1f%%")
plt.title("Payment Mode Distribution")
plt.ylabel("")
plt.show()
```

```python
df["VisitDate"] = pd.to_datetime(df["VisitDate"])
```

```python
monthly = df.groupby(df["VisitDate"].dt.to_period("M")).size()

plt.figure(figsize=(10,5))
monthly.plot(kind="line", marker="o")
plt.title("Monthly Patient Visits")
plt.xlabel("Month")
plt.ylabel("Patients")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["TreatmentCost"], df["TotalBill"])
plt.title("Treatment Cost vs Total Bill")
plt.xlabel("Treatment Cost")
plt.ylabel("Total Bill")
plt.show()
```

```python
satisfaction = df.groupby("Department")["PatientSatisfaction"].mean()

plt.figure(figsize=(8,5))
satisfaction.plot(kind="bar")
plt.title("Average Patient Satisfaction by Department")
plt.xlabel("Department")
plt.ylabel("Average Satisfaction")
plt.show()
```
